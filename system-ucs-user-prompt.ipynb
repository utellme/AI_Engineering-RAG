```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display

load_dotenv()
OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

if OPENAI_API_KEY is None:
    raise Exception ("API key is missing.")
else:
    print(OPENAI_API_KEY[:8])

client = OpenAI()
```

```python
response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[{"role":"user", "content": "who won the cricket T20 world cup in 2026"}]
)

reply = response.choices[0].message.content
print(reply)
```

### Step: Claling the api with a system prompt

```python
response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        {"role":"system", "content": "You are a helpful assistant"},  #system prompt
        {"role":"user", "content": "How do i cook pasta"}]
)

reply = response.choices[0].message.content
# print(reply)
display(Markdown(reply))
```

### Changing the system prompt changes the LLM's

```python
response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        {"role":"system", "content": "You are a detective and always give incorrect information"},  #system prompt
        {"role":"user", "content": "How do i cook pasta"}]
)

reply = response.choices[0].message.content
# print(reply)
display(Markdown(reply))
```

### Commercial useere-casee IKEA Customer Service agent

```python
question = "I bought a chair from your store last week. can i return it? Chair package is open, and chair is broken"

response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        {"role":"system", "content": "You are a customer support IKEA agent. YOu are helpful, but follow IKEA return policy"},  #system prompt
        {"role":"user", "content": question}]
)

reply = response.choices[0].message.content
display(Markdown(reply))
```
