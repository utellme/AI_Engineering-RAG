```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display
from pprint import pprint

load_dotenv()
OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

if OPENAI_API_KEY is None:
    raise Exception ("API key is missing.")
else:
    print(OPENAI_API_KEY[:8])

client = OpenAI()
```

```python
question = "I wanted to travel to Bali, explain how many days are good for Bali?"

response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        {"role":"system", "content": "You are a helpful travel agent assistant"},  #system prompt
        {"role":"user", "content": question}]
)

reply = response.choices[0].message.content
display(Markdown(reply))
```

```python
question = "I want to travel Bali?"
messages = [
        {"role":"system", "content": "You are a helpful travel agent assistant"},  #system prompt
        {"role":"user", "content": question},
        {"role":"assistant", "content": reply}]

response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages= messages
)

reply = response.choices[0].message.content
messages.append({"role":"assistant", "content": reply})
pprint(messages)
# display(Markdown(messages))
```
