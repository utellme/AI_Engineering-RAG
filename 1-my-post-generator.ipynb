### step1: ijport library and API keys

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display

load_dotenv()
OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

if OPENAI_API_KEY is None:
    raise Exception ("API key is missing.")
else:
    print(OPENAI_API_KEY[:8])

client = OpenAI()
```

### Step2: Searching the web

```python
client = OpenAI()

completions = client.chat.completions.create(
    model="gpt-4o-mini-search-preview",
    messages = [
         # {"role":"system", "content": "You are a helpful assistant"},  #system prompt
        {"role":"user", "content": "what is today's datee? Tell me 3 latest fun news latest in the world"}
    ],
    max_tokens=500
)
```

```python
search_results = completions.choices[0].message.content
display(Markdown(search_results))
```

```python
### Step3: Summarize and write social media post
```

```python
system_message = """
You are a helpful assistant who creates amazing LinkedIn post about AI based latest fun newes. Make concise and add a thought proviling quesiton at the end to ecourage comments
"""

prompt = f"""
    The folling text contains top 3 latest fun stories from AI. Please write my linkedin post baseed on this information.
    {search_results}
"""

completion = client.chat.completions.create(
    model="gpt-4o-mini",
    messages = [
        {"role":"system", "content": system_message},  #system prompt
        {"role":"user", "content": prompt}
    ],
    max_tokens=500,
    temperature=0.7
)
chat_response = completion.choices[0].message.content
```
