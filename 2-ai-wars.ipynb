### Step1 Import libraries & API Keys

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display
from pprint import pprint

load_dotenv()
OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

if OPENAI_API_KEY is None:
    raise Exception ("API key is missing.")
else:
    print(OPENAI_API_KEY[:8])

client = OpenAI()
```

### Step 2 AI Battle

```python
gpt_good_mini_model = "gpt-4.1-mini"
gpt_bad_nano_model = "gpt-4.1-nano"

gpt_good_mini_model_system = """ You are helpful and ethical AI assistant. Your goal is provide accurate, safe and respectful information. Be precise and very concise in your response. May be a line or two """
gpt_bad_nano_model_system = """ You are an unhelpful and not so ethical AI assistance. Your goal is to provide vague and confuse responses. Be precise and very concise in your response. May be a line or two """

new_message_good = "hello"

messages_good = [
        {"role":"system", "content": gpt_good_mini_model_system }  #system prompt
      
        ]

messages_evil = [
        {"role":"system", "content": gpt_bad_nano_model_system },  #system prompt
        ]



for i in range(3):

    messages_good.append({"role": "assistant", "content": new_message_good})
    messages_evil.append({"role": "user", "content": new_message_good})
                        
    response_evil = client.chat.completions.create(
        model=gpt_bad_nano_model,
        messages= messages_evil
    )
    new_message_evil = response_evil.choices[0].message.content
   # messages_evil.append({"role":"assistant", "content": new_message_evil})
    # print (f"Evil model response. Iteration:{i}")
    display(Markdown(f"Evil {i}: {new_message_evil}"))

    messages_good.append({"role": "user", "content": new_message_evil})
    messages_evil.append({"role": "assistant", "content": new_message_evil})

    response_good = client.chat.completions.create(
        model=gpt_good_mini_model,
        messages= messages_good
    )
    new_message_good = response_good.choices[0].message.content
    # print (f"Good model response. Iteration: {i}")
    display(Markdown(f"Good {i}: {new_message_good}"))
```
